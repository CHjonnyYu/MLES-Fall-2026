## 2. Reconstruct the attention calculation

Solve before running: scores [1,0,1]/sqrt(2), values [[10,0],[0,10],[10,10]].


```python
scores=np.array([1.,0.,1.])/np.sqrt(2)
weights=np.exp(scores-scores.max()); weights/=weights.sum()
values=np.array([[10.,0.],[0.,10.],[10.,10.]])
context=weights@values
print("Weights:",weights,"Context:",context)
assert np.isclose(weights.sum(),1)
```

**Checkpoint.** You should explain scores, softmax, and weighted retrieval as three distinct operations.
