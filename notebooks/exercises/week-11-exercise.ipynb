## 2. Compute metrics from counts

Prepare for a written question: 18 true positives, 2 false negatives, 8 false positives, and 72 true negatives. Compute by hand before running.


```python
tp,fn,fp,tn=18,2,8,72
summary={"recall":tp/(tp+fn),"precision":tp/(tp+fp),
         "specificity":tn/(tn+fp),"accuracy":(tp+tn)/(tp+fn+fp+tn)}
print(summary)
assert np.isclose(summary["recall"],.9)
```

**Checkpoint.** Explain why precision is not recall. Recall is 0.9; the denominators refer to different populations.


## 3. Audit a timeline for leakage

A forecast at index t may use indices strictly below t. Write and inspect a counterexample with one future feature.


```python
origin=100
good=np.arange(76,100); bad=np.arange(77,101)
print("Safe history:",bool((good<origin).all()))
print("Unsafe history:",bool((bad<origin).all()))
assert good.max()<origin and not (bad.max()<origin)
```

**Checkpoint.** The unsafe window contains index 100, the target time. Identify the error before discussing model accuracy.
