## 1. Set up and record the environment

Keep the seed fixed for the first run. Record actual outputs and package versions rather than copying a claimed performance number.


```python
import sys, subprocess, importlib.util, platform, json, random, hashlib
from pathlib import Path
required = {"numpy": "numpy", "matplotlib": "matplotlib", "sklearn": "scikit-learn"}
missing = [package for module, package in required.items() if importlib.util.find_spec(module) is None]
if missing:
    subprocess.check_call([sys.executable, "-m", "pip", "install", *missing])
import numpy as np
import matplotlib.pyplot as plt
SEED = 2026
random.seed(SEED)
np.random.seed(SEED)
rng = np.random.default_rng(SEED)
ARTIFACTS = Path("artifacts")
ARTIFACTS.mkdir(exist_ok=True)
print("Python:", platform.python_version(), "NumPy:", np.__version__, "Seed:", SEED)
```

## 2. Package an independently interpretable result

Create a small original artifact and an explicit claim. A result card should distinguish measured behavior from broader ambitions.


```python
signal=rng.normal(size=100)
card={"task":"mean of a generated normal sample","seed":SEED,"n":100,
      "observed_mean":float(signal.mean()),"units":"dimensionless synthetic",
      "limitation":"one random sample; no real engineering measurement"}
```

**Checkpoint.** The sample mean will not generally equal zero exactly; zero is the population expectation.


## 3. Verify data identity

Save the array with a hash based on its bytes and reload it. This checks artifact integrity, not scientific validity.


```python
path=ARTIFACTS/"final-synthetic.npy"; np.save(path,signal)
digest=hashlib.sha256(path.read_bytes()).hexdigest()
restored=np.load(path)
assert np.array_equal(signal,restored)
card["file_sha256"]=digest
print(card)
```
