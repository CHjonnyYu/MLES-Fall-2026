## 1. Set up and record the environment

Keep the seed fixed for the first run. Record actual outputs and package versions rather than copying a claimed performance number.


```python
import sys, subprocess, importlib.util, platform, json, random, hashlib
from pathlib import Path
required = {"numpy": "numpy", "matplotlib": "matplotlib", "sklearn": "scikit-learn"}
missing = [package for module, package in required.items() if importlib.util.find_spec(module) is None]
if missing:
    subprocess.check_call([sys.executable, "-m", "pip", "install", *missing])
import numpy as np
import matplotlib.pyplot as plt
SEED = 2026
random.seed(SEED)
np.random.seed(SEED)
rng = np.random.default_rng(SEED)
ARTIFACTS = Path("artifacts")
ARTIFACTS.mkdir(exist_ok=True)
print("Python:", platform.python_version(), "NumPy:", np.__version__, "Seed:", SEED)
```

**Checkpoint.** The environment cell must finish before any analysis cell. Restart and run all if a variable is missing.


## 2. Create a traceable measurement table

Each row is a synthetic observation, not a real laboratory result. Keep sample identity separate from measured values.


```python
sample_id=np.arange(1,61)
temperature=25+rng.normal(0,2,60)
response=2.0*temperature+rng.normal(0,1,60)
data=np.column_stack([sample_id,temperature,response])
print("Columns: sample_id, synthetic temperature, synthetic response")
print(data[:5])
```

**Checkpoint.** There are 60 rows and 3 columns. The sample ID must never be treated as a physical measurement.


## 3. Inspect and correct a missing value

Introduce a missing temperature at a known row. Preserve the original array, and create a cleaned copy. Median imputation is a declared modeling assumption.


```python
raw=data.copy(); raw[7,1]=np.nan
clean=raw.copy(); median=np.nanmedian(raw[:,1]); clean[np.isnan(clean[:,1]),1]=median
print("Missing raw values:",np.isnan(raw).sum(),"After cleaning:",np.isnan(clean).sum())
assert np.isfinite(clean).all()
```

**Checkpoint.** The raw table still contains one missing value. Explain why silently modifying raw measurements would break provenance.


## 4. Visualize a claim

Plot the relationship and label axes with synthetic units. Correlation in this generated example is not a causal finding about actual materials.


```python
plt.scatter(clean[:,1],clean[:,2]); plt.xlabel("Synthetic temperature units")
plt.ylabel("Synthetic response units"); plt.title("Synthetic teaching data"); plt.show()
summary={"rows":len(clean),"mean_response":float(clean[:,2].mean()),"imputation":"median temperature"}
print(summary)
```

**Checkpoint.** The figure should show an increasing relationship. Report row count and the declared cleaning rule.

### Task 2: Identify data without claiming authenticity

Hash the original and cleaned tables. Predict whether the hashes match. Change one measurement and explain what a changed hash does and does not establish.

**Steps**
1. Write your prediction in `responses['task2_prediction']` before running.
2. Run the starter cell; read every printed quantity and retain the observation.
3. Make the controlled comparison requested above. Record its configuration in `responses['controlled_change']`.

**Checkpoint:** The hashes differ. A hash identifies bytes; it does not prove that a measurement is genuine.

```python
raw_digest=hashlib.sha256(raw_demo.tobytes()).hexdigest()
clean_digest=hashlib.sha256(clean_demo.tobytes()).hexdigest()
observations['task2']={'raw_sha256':raw_digest,'clean_sha256':clean_digest,'same_bytes':raw_digest==clean_digest}
print(observations['task2'])
```
